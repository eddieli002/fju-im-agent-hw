### 第 2 格　讀出 `name` 與 `description`

這一格只做一件事：**把 Frontmatter 解析出來印給你看。**

模型做觸發判定的時候，讀得到的就只有這兩行——`SKILL.md` 底下的
`## 執行原則`、`## 邊界條件`、`## 完成條件` 它都看不到。

**預期輸出**：`name` 與 `description` 各一行，以及 `description` 的字數。

```python
def 讀frontmatter(路徑):
    """把 --- 包起來的那一段解析成 dict。只處理 `鍵: 值` 這一種，夠用。"""
    文 = open(路徑, encoding="utf-8").read()
    行 = 文.split("\n")

    assert 行 and 行[0].strip() == "---", (
        "第 1 行不是 ---（要三個半形連字號，不是全形破折號）。"
        "實際讀到：%r" % (行[0][:20] if 行 else ""))

    try:
        結束 = next(i for i in range(1, len(行)) if 行[i].strip() == "---")
    except StopIteration:
        raise AssertionError("找不到第二個 ---，Frontmatter 沒有收尾。")

    fm = {}
    for l in 行[1:結束]:
        if not l.strip():
            continue
        assert ":" in l, "Frontmatter 這一行沒有冒號：%r" % l
        k, v = l.split(":", 1)
        fm[k.strip()] = v.strip()

    for 必要 in ("name", "description"):
        assert 必要 in fm, (
            "Frontmatter 缺 %s。目前只有：%s" % (必要, "、".join(fm) or "（空）"))
    return fm


FM = 讀frontmatter(SKILL_PATH)
NAME = FM["name"]
DESC = FM["description"]

print("name")
print("    %s" % NAME)
print()
print("description（%d 字）" % len(DESC))
print("    %s" % DESC)
print()
if NAME != SKILL_NAME:
    print("⚠ name（%s）與資料夾名稱（%s）不一樣。" % (NAME, SKILL_NAME))
    print("  不影響這次判定，但照操作說明做的話兩個應該相同。")
else:
    print("✓ name 與資料夾名稱相同")
```

```python
import re

# ── 判準（由 填答區.py 產生，不要在這裡改數字）──────────────────
區塊順序 = ['何時載入 references', '執行原則', '邊界條件', '完成條件']
佔位記號 = '【'
上限 = {'SKILL行數': 40, 'references行數': 20, 'description字數': 120, '自寫每條字數': 40, 'SKILL_token': 600}
執行原則條數 = (3, 4)

S = open(SKILL_PATH, encoding="utf-8").read()
R = open(refs[0], encoding="utf-8").read()


def 節內(標題, 樣式):
    m = re.search(r"^## %s$(.*?)(?=^## |\Z)" % re.escape(標題), S, re.M | re.S)
    return re.findall(樣式, m.group(1), re.M) if m else []


h2 = re.findall(r"^## (.+)$", S, re.M)
原則 = 節內("執行原則", r"^\d+\.\s*(.+)$")
邊界 = 節內("邊界條件", r"^-\s*(.+)$")
完成 = 節內("完成條件", r"^-\s*\[ \]\s*(.+)$")
載入節 = 節內("何時載入 references", r"^-\s*(.+)$")
自寫 = [s for s in 原則[2:] if 佔位記號 not in s]
留著 = [s for s in 原則 if 佔位記號 in s]

# 這七條就是「操作說明」第一節列的那七條，一條不多一條不少。
七條規定 = [
    ("檔案剛好 2 個", "SKILL.md ＋ %d 份 refs" % len(refs), len(refs) == 1),
    ("4 個 ## 區塊且順序正確", "／".join(h2), h2 == 區塊順序),
    ("指名 references 裡的檔名", "有" if any("references/" in s for s in 載入節)
     else "沒有", any("references/" in s for s in 載入節)),
    ("執行原則 3 或 4 條", "%d 條" % len(原則),
     執行原則條數[0] <= len(原則) <= 執行原則條數[1]),
    ("自己寫的每條 %d 字以內" % 上限["自寫每條字數"],
     "／".join(str(len(s)) for s in 自寫) or "（沒有自寫條目）",
     bool(自寫) and all(len(s) <= 上限["自寫每條字數"] for s in 自寫)),
    ("SKILL.md %d 行以內" % 上限["SKILL行數"], "%d 行" % len(S.splitlines()),
     len(S.splitlines()) <= 上限["SKILL行數"]),
    ("references %d 行以內" % 上限["references行數"],
     "%d 行" % len(R.splitlines()), len(R.splitlines()) <= 上限["references行數"]),
]

# 這三條不在那七條裡面，因為它們是「照抄的部分」與「步驟 5」——
# 照操作說明做就一定會過。列出來是為了在沒過的時候講清楚是哪裡沒做。
另外確認 = [
    ("邊界條件至少 1 條", "%d 條" % len(邊界), len(邊界) >= 1),
    ("完成條件至少 2 條", "%d 條" % len(完成), len(完成) >= 2),
    ("佔位行都換掉了（步驟 5）",
     "還有 %d 行" % len(留著) if 留著 else "已換掉", not 留著),
]

壞 = 0
print("=" * 60)
print("  操作說明第一節的七條規定")
print("=" * 60)
for 項, 值, 過 in 七條規定:
    if not 過:
        壞 += 1
    print("  %s %-26s %s" % ("✓" if 過 else "✗", 項, 值))
print()
print("=" * 60)
print("  另外確認（照抄的部分，照操作說明做就會過）")
print("=" * 60)
for 項, 值, 過 in 另外確認:
    if not 過:
        壞 += 1
    print("  %s %-26s %s" % ("✓" if 過 else "✗", 項, 值))
print("=" * 60)
if 壞:
    print("  有 %d 條沒過。回去改，改完重新壓 zip、從第 1 格重跑。" % 壞)
else:
    print("  全部通過，可以往下做。")
```

### 第 6 格　載入模型

**這一格開始需要 GPU。** 還在 CPU 的話先換：
**執行階段 → 變更執行階段類型 → T4 GPU**，然後從第 1 格重跑。

第一次執行要下載模型，大約 3–5 分鐘。

**預期輸出**：`模型已載入，route() 已定義`。

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL = "Qwen/Qwen2.5-3B-Instruct"
tok = AutoTokenizer.from_pretrained(MODEL)
model = AutoModelForCausalLM.from_pretrained(MODEL, dtype=torch.float16, device_map="auto")
model.eval()


def gen(prompt, max_new_tokens=32):
    text = tok.apply_chat_template(
        [{"role": "user", "content": prompt}],
        tokenize=False, add_generation_prompt=True)
    ids = tok(text, return_tensors="pt").to(model.device)
    with torch.no_grad():
        out = model.generate(**ids, max_new_tokens=max_new_tokens, do_sample=False)
    return tok.decode(out[0][ids["input_ids"].shape[1]:], skip_special_tokens=True).strip()


def route(pool, question):
    listing = "\n".join(f"- {n}: {d}" for n, d in pool)
    prompt = (f"以下是可用的技能清單，每一個只給名稱與描述：\n\n{listing}\n\n"
             f"使用者說：「{question}」\n\n"
             "請只回答最適合的技能名稱；若沒有任何一個適合，只回答 none。不要解釋。")
    ans = gen(prompt)
    picked = ans.strip().split()[0].strip(".,:：「」") if ans.strip() else "(空)"
    return picked, ans


print("模型已載入，route() 已定義")
```
